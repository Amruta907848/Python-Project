 #### Shortest Path Finding System

```python
import tkinter as tk
from tkinter import messagebox
import heapq
import math
```

```python
INF = float("inf")
```

```python
BG_COLOR = "#F5F7FA"
SIDEBAR_COLOR = "#EAF1F8"
HEADER_COLOR = "#DCE8F5"
CARD_COLOR = "#FFFFFF"

PRIMARY_COLOR = "#4F81BD"
SUCCESS_COLOR = "#5B9BD5"
WARNING_COLOR = "#E6A23C"
DANGER_COLOR = "#D96C6C"
PURPLE_COLOR = "#8064A2"

TEXT_COLOR = "#263238"
SECONDARY_TEXT = "#607080"

WHITE = "#FFFFFF"
BORDER_COLOR = "#D5DEE8"
INPUT_BORDER = "#C5D1DE"
GRAPH_BG = "#FBFCFE"
```

```python
class ShortestPathApp:

    def __init__(self, root):

        self.root = root

        self.root.title(
            "Shortest Path Finding System"
        )

        self.root.geometry("1150x720")
        self.root.minsize(1050, 680)

        self.root.configure(
            bg=BG_COLOR
        )

       
        self.graph = {}


        header = tk.Frame(
            root,
            bg=HEADER_COLOR,
            height=85
        )

        header.pack(
            fill="x"
        )

        header.pack_propagate(False)

        tk.Label(
            header,
            text="SHORTEST PATH FINDING SYSTEM",
            font=("Segoe UI", 21, "bold"),
            bg=HEADER_COLOR,
            fg=TEXT_COLOR
        ).pack(
            pady=(13, 2)
        )

        tk.Label(
            header,
            text="Dijkstra's Algorithm  &  All Pairs Shortest Path",
            font=("Segoe UI", 10),
            bg=HEADER_COLOR,
            fg=SECONDARY_TEXT
        ).pack()

        main = tk.Frame(
            root,
            bg=BG_COLOR
        )

        main.pack(
            fill="both",
            expand=True,
            padx=15,
            pady=15
        )



        left = tk.Frame(
            main,
            bg=SIDEBAR_COLOR,
            width=320,
            height=665,
            highlightbackground=BORDER_COLOR,
            highlightthickness=1
        )

        left.pack(
            side="left",
            fill="y",
            padx=(0, 15)
        )


        left.pack_propagate(False)


        tk.Label(
            left,
            text="GRAPH INPUT",
            font=("Segoe UI", 13, "bold"),
            bg=SIDEBAR_COLOR,
            fg=TEXT_COLOR
        ).pack(
            pady=(12, 7)
        )



        self.section_label(
            left,
            "1. Add Vertex"
        )

        self.label_hint(
            left,
            "Vertex"
        )

        self.vertex_entry = self.entry(
            left
        )

        self.button(
            left,
            "Add Vertex",
            PRIMARY_COLOR,
            self.add_vertex
        )



        tk.Label(
            left,
            text="Added Vertices",
            font=("Segoe UI", 9, "bold"),
            bg=SIDEBAR_COLOR,
            fg=SECONDARY_TEXT
        ).pack(
            anchor="w",
            padx=20,
            pady=(5, 2)
        )

        list_frame = tk.Frame(
            left,
            bg=SIDEBAR_COLOR
        )

        list_frame.pack(
            padx=20,
            fill="x"
        )

        self.vertex_listbox = tk.Listbox(
            list_frame,
            height=2,
            font=("Segoe UI", 9),
            bg=WHITE,
            fg=TEXT_COLOR,
            selectbackground=PRIMARY_COLOR,
            selectforeground=WHITE,
            relief="flat",
            highlightbackground=INPUT_BORDER,
            highlightthickness=1
        )

        self.vertex_listbox.pack(
            fill="x"
        )


        self.section_label(
            left,
            "2. Add Edge"
        )

        self.label_hint(
            left,
            "Source"
        )

        self.edge_source = self.entry(
            left
        )

        self.label_hint(
            left,
            "Destination"
        )

        self.edge_destination = self.entry(
            left
        )

        self.label_hint(
            left,
            "Weight"
        )

        self.edge_weight = self.entry(
            left
        )

        self.button(
            left,
            "Add Edge",
            SUCCESS_COLOR,
            self.add_edge
        )



        self.section_label(
            left,
            "3. Find Shortest Path"
        )

        self.label_hint(
            left,
            "Source"
        )

        self.source_entry = self.entry(
            left
        )

        self.label_hint(
            left,
            "Destination"
        )

        self.destination_entry = self.entry(
            left
        )


        self.button(
            left,
            "Run Dijkstra",
            PRIMARY_COLOR,
            self.run_dijkstra
        )



        self.button(
            left,
            "All Pairs Shortest Path",
            PURPLE_COLOR,
            self.run_floyd
        )


        button_frame = tk.Frame(
            left,
            bg=SIDEBAR_COLOR
        )

        button_frame.pack(
            fill="x",
            padx=20,
            pady=2
        )


        tk.Button(
            button_frame,
            text="Clear Input",
            command=self.clear_input,
            font=("Segoe UI", 8, "bold"),
            bg=WARNING_COLOR,
            fg=WHITE,
            activebackground="#D18B25",
            activeforeground=WHITE,
            relief="flat",
            cursor="hand2",
            padx=4,
            pady=5
        ).pack(
            side="left",
            fill="x",
            expand=True,
            padx=(0, 3)
        )


        tk.Button(
            button_frame,
            text="Clear All",
            command=self.clear_all,
            font=("Segoe UI", 8, "bold"),
            bg=DANGER_COLOR,
            fg=WHITE,
            activebackground="#C45151",
            activeforeground=WHITE,
            relief="flat",
            cursor="hand2",
            padx=4,
            pady=5
        ).pack(
            side="right",
            fill="x",
            expand=True,
            padx=(3, 0)
        )



        right = tk.Frame(
            main,
            bg=BG_COLOR
        )

        right.pack(
            side="right",
            fill="both",
            expand=True
        )

        graph_card = tk.Frame(
            right,
            bg=CARD_COLOR,
            highlightbackground=BORDER_COLOR,
            highlightthickness=1
        )

        graph_card.pack(
            fill="both",
            expand=True
        )


        tk.Label(
            graph_card,
            text="Graph Visualization",
            font=("Segoe UI", 13, "bold"),
            bg=CARD_COLOR,
            fg=TEXT_COLOR
        ).pack(
            anchor="w",
            padx=15,
            pady=9
        )


        self.canvas = tk.Canvas(
            graph_card,
            bg=GRAPH_BG,
            highlightthickness=0
        )

        self.canvas.pack(
            fill="both",
            expand=True,
            padx=10,
            pady=(0, 10)
        )


        result_card = tk.Frame(
            right,
            bg=CARD_COLOR,
            highlightbackground=BORDER_COLOR,
            highlightthickness=1
        )

        result_card.pack(
            fill="both",
            expand=True,
            pady=(15, 0)
        )


        tk.Label(
            result_card,
            text="Algorithm Result",
            font=("Segoe UI", 13, "bold"),
            bg=CARD_COLOR,
            fg=TEXT_COLOR
        ).pack(
            anchor="w",
            padx=15,
            pady=9
        )


        text_frame = tk.Frame(
            result_card,
            bg=CARD_COLOR
        )

        text_frame.pack(
            fill="both",
            expand=True,
            padx=10,
            pady=(0, 10)
        )


        self.result_text = tk.Text(
            text_frame,
            font=("Consolas", 10),
            bg="#F8FAFC",
            fg=TEXT_COLOR,
            insertbackground=TEXT_COLOR,
            relief="flat",
            highlightbackground=INPUT_BORDER,
            highlightthickness=1,
            wrap="none"
        )

        self.result_text.pack(
            side="left",
            fill="both",
            expand=True
        )


        scrollbar = tk.Scrollbar(
            text_frame,
            command=self.result_text.yview
        )

        scrollbar.pack(
            side="right",
            fill="y"
        )


        self.result_text.config(
            yscrollcommand=scrollbar.set
        )


        # Initial graph message
        self.show_graph()



    def section_label(
        self,
        parent,
        text
    ):

        tk.Label(
            parent,
            text=text,
            font=("Segoe UI", 10, "bold"),
            bg=SIDEBAR_COLOR,
            fg=PRIMARY_COLOR
        ).pack(
            anchor="w",
            padx=20,
            pady=(4, 1)
        )



    def small_info(
        self,
        parent,
        text
    ):

        tk.Label(
            parent,
            text=text,
            font=("Segoe UI", 7),
            bg=SIDEBAR_COLOR,
            fg=SECONDARY_TEXT
        ).pack(
            anchor="w",
            padx=20,
            pady=(0, 1)
        )


    def entry(
        self,
        parent
    ):

        entry = tk.Entry(
            parent,
            font=("Segoe UI", 9),
            bg=WHITE,
            fg=TEXT_COLOR,
            insertbackground=TEXT_COLOR,
            relief="flat",
            highlightbackground=INPUT_BORDER,
            highlightcolor=PRIMARY_COLOR,
            highlightthickness=1
        )

        entry.pack(
            fill="x",
            padx=20,
            pady=0,
            ipady=3
        )

        return entry


    def label_hint(
        self,
        parent,
        text
    ):

        label = tk.Label(
            parent,
            text=text,
            font=("Segoe UI", 7),
            bg=SIDEBAR_COLOR,
            fg=SECONDARY_TEXT
        )

        label.pack(
            anchor="w",
            padx=22,
            pady=(0, 0)
        )

        return label



    def button(
        self,
        parent,
        text,
        color,
        command
    ):

        tk.Button(
            parent,
            text=text,
            command=command,
            font=("Segoe UI", 8, "bold"),
            bg=color,
            fg=WHITE,
            activebackground=color,
            activeforeground=WHITE,
            relief="flat",
            cursor="hand2",
            pady=4
        ).pack(
            fill="x",
            padx=20,
            pady=1
        )



    def add_vertex(self):

        vertex = (
            self.vertex_entry
            .get()
            .strip()
        )


        if vertex == "":

            messagebox.showwarning(
                "Input Required",
                "Please enter a vertex name.\n\n"
                "Example: A"
            )

            return


        if vertex in self.graph:

            messagebox.showwarning(
                "Duplicate Vertex",
                f"Vertex '{vertex}' already exists."
            )

            return


        self.graph[vertex] = {}


        self.vertex_listbox.insert(
            tk.END,
            vertex
        )


        self.vertex_entry.delete(
            0,
            tk.END
        )


        self.show_graph()



    def add_edge(self):

        source = (
            self.edge_source
            .get()
            .strip()
        )

        destination = (
            self.edge_destination
            .get()
            .strip()
        )

        weight_text = (
            self.edge_weight
            .get()
            .strip()
        )


      
        if source == "":

            messagebox.showerror(
                "Source Required",
                "Please enter Source vertex.\n\n"
                "Example: A"
            )

            return


        if source not in self.graph:

            messagebox.showerror(
                "Invalid Source",
                f"Vertex '{source}' does not exist.\n\n"
                "Please add the vertex first."
            )

            return


     
        if destination == "":

            messagebox.showerror(
                "Destination Required",
                "Please enter Destination vertex.\n\n"
                "Example: B"
            )

            return


        if destination not in self.graph:

            messagebox.showerror(
                "Invalid Destination",
                f"Vertex '{destination}' does not exist.\n\n"
                "Please add the vertex first."
            )

            return


     
        if weight_text == "":

            messagebox.showerror(
                "Weight Required",
                "Please enter edge weight.\n\n"
                "Example: 5"
            )

            return


        try:

            weight = float(
                weight_text
            )

        except ValueError:

            messagebox.showerror(
                "Invalid Weight",
                "Weight must be a number.\n\n"
                "Example: 5 or 2.5"
            )

            return


        # Dijkstra requires non-negative weights
        if weight < 0:

            messagebox.showerror(
                "Invalid Weight",
                "Negative weights are not allowed.\n\n"
                "Please enter 0 or a positive number."
            )

            return


        # Undirected edge: store the connection in both directions
        # so the shortest path can travel either way.
        self.graph[source][destination] = weight
        self.graph[destination][source] = weight


        # Clear edge inputs
        self.edge_source.delete(
            0,
            tk.END
        )

        self.edge_destination.delete(
            0,
            tk.END
        )

        self.edge_weight.delete(
            0,
            tk.END
        )


        self.show_graph()


    # ========================================================
    # DIJKSTRA ALGORITHM
    # ========================================================

    def dijkstra(
        self,
        source
    ):

        distance = {
            vertex: INF
            for vertex in self.graph
        }


        previous = {
            vertex: None
            for vertex in self.graph
        }


        distance[source] = 0


        priority_queue = [
            (0, source)
        ]


        while priority_queue:

            current_distance, current_vertex = \
                heapq.heappop(
                    priority_queue
                )


            if (
                current_distance
                >
                distance[current_vertex]
            ):

                continue


            for neighbour, weight in \
                    self.graph[
                        current_vertex
                    ].items():

                new_distance = (
                    current_distance
                    + weight
                )


                if (
                    new_distance
                    <
                    distance[neighbour]
                ):

                    distance[neighbour] = \
                        new_distance

                    previous[neighbour] = \
                        current_vertex


                    heapq.heappush(
                        priority_queue,
                        (
                            new_distance,
                            neighbour
                        )
                    )


        return distance, previous


    # ========================================================
    # GET SHORTEST PATH
    # ========================================================

    def get_path(
        self,
        previous,
        source,
        destination
    ):

        path = []

        current = destination


        while current is not None:

            path.append(
                current
            )

            current = previous[
                current
            ]


        path.reverse()


        if (
            len(path) == 0
            or
            path[0] != source
        ):

            return []


        return path


    # ========================================================
    # RUN DIJKSTRA
    # ========================================================

    def run_dijkstra(self):

        source = (
            self.source_entry
            .get()
            .strip()
        )

        destination = (
            self.destination_entry
            .get()
            .strip()
        )


        # Graph check
        if len(self.graph) == 0:

            messagebox.showwarning(
                "No Graph",
                "Please add vertices and edges first."
            )

            return


        # Source check
        if source == "":

            messagebox.showerror(
                "Source Required",
                "Please enter Source vertex.\n\n"
                "Example: A"
            )

            return


        if source not in self.graph:

            messagebox.showerror(
                "Invalid Source",
                f"Vertex '{source}' does not exist."
            )

            return


        # Destination check
        if destination == "":

            messagebox.showerror(
                "Destination Required",
                "Please enter Destination vertex.\n\n"
                "Example: D"
            )

            return


        if destination not in self.graph:

            messagebox.showerror(
                "Invalid Destination",
                f"Vertex '{destination}' does not exist."
            )

            return


        # Run algorithm
        distance, previous = \
            self.dijkstra(
                source
            )


        # Clear result
        self.result_text.delete(
            "1.0",
            tk.END
        )


        # Header
        self.result_text.insert(
            tk.END,
            "================================================\n"
        )

        self.result_text.insert(
            tk.END,
            "             DIJKSTRA'S ALGORITHM\n"
        )

        self.result_text.insert(
            tk.END,
            "          SINGLE SOURCE SHORTEST PATH\n"
        )

        self.result_text.insert(
            tk.END,
            "================================================\n\n"
        )


        self.result_text.insert(
            tk.END,
            f"Source      : {source}\n"
        )

        self.result_text.insert(
            tk.END,
            f"Destination : {destination}\n\n"
        )


        # Distances
        self.result_text.insert(
            tk.END,
            "SHORTEST DISTANCES\n"
        )

        self.result_text.insert(
            tk.END,
            "------------------------------------------------\n"
        )


        for vertex in self.graph:

            if distance[vertex] == INF:

                value = "No path"

            else:

                value = self.format_number(
                    distance[vertex]
                )


            self.result_text.insert(
                tk.END,
                f"{vertex:<10} : {value}\n"
            )


        # Destination
        self.result_text.insert(
            tk.END,
            "\n"
        )


        if distance[destination] == INF:

            self.result_text.insert(
                tk.END,
                "NO PATH EXISTS\n"
            )

        else:

            path = self.get_path(
                previous,
                source,
                destination
            )


            self.result_text.insert(
                tk.END,
                "SHORTEST PATH\n"
            )

            self.result_text.insert(
                tk.END,
                "------------------------------------------------\n"
            )


            self.result_text.insert(
                tk.END,
                " -> ".join(path)
            )


            self.result_text.insert(
                tk.END,
                "\n\n"
            )


            self.result_text.insert(
                tk.END,
                "Total Distance : "
            )


            self.result_text.insert(
                tk.END,
                self.format_number(
                    distance[destination]
                )
            )


    # ========================================================
    # FLOYD-WARSHALL / ALL PAIRS SHORTEST PATH
    # ========================================================

    def run_floyd(self):

        # Graph check
        if len(self.graph) == 0:

            messagebox.showwarning(
                "No Graph",
                "Please add vertices and edges first."
            )

            return


        # Get vertices
        vertices = list(
            self.graph.keys()
        )


        n = len(vertices)


        # ====================================================
        # CREATE DISTANCE MATRIX
       

        dist = [
            [
                INF
                for _ in range(n)
            ]
            for _ in range(n)
        ]


        # Distance from vertex to itself
        for i in range(n):

            dist[i][i] = 0


        # Put edge weights in matrix
        for i in range(n):

            source = vertices[i]

            for destination, weight in \
                    self.graph[source].items():

                j = vertices.index(
                    destination
                )

                dist[i][j] = weight


        # CLEAR RESULT
        

        self.result_text.delete(
            "1.0",
            tk.END
        )


    
        # HEADER
      

        self.result_text.insert(
            tk.END,
            "============================================================\n"
        )

        self.result_text.insert(
            tk.END,
            "             FLOYD-WARSHALL ALGORITHM\n"
        )

        self.result_text.insert(
            tk.END,
            "              ALL PAIRS SHORTEST PATH\n"
        )

        self.result_text.insert(
            tk.END,
            "============================================================\n\n"
        )


        self.result_text.insert(
            tk.END,
            "This algorithm finds the shortest distance\n"
            "between every pair of vertices.\n\n"
        )


        
        # INITIAL MATRIX D0
        

        self.print_matrix(
            vertices,
            dist,
            "D0 - INITIAL MATRIX"
        )


       
        # FLOYD-WARSHALL STEPS

        for k in range(n):

            for i in range(n):

                for j in range(n):

                    if (
                        dist[i][k] != INF
                        and
                        dist[k][j] != INF
                    ):

                        new_distance = (
                            dist[i][k]
                            + dist[k][j]
                        )


                        if (
                            new_distance
                            <
                            dist[i][j]
                        ):

                            dist[i][j] = \
                                new_distance


            # Print intermediate matrix
            self.print_matrix(
                vertices,
                dist,
                f"D{k + 1} - USING VERTEX "
                f"{vertices[k]}"
            )


        # ====================================================
        # FINAL MATRIX
        # ====================================================

        self.print_matrix(
            vertices,
            dist,
            "FINAL SHORTEST DISTANCE MATRIX"
        )


    # ========================================================
    # PRINT MATRIX
    # ========================================================

    def print_matrix(
        self,
        vertices,
        matrix,
        title
    ):

        self.result_text.insert(
            tk.END,
            "\n"
            + "=" * 60
            + "\n"
        )


        self.result_text.insert(
            tk.END,
            title
            + "\n"
        )


        self.result_text.insert(
            tk.END,
            "=" * 60
            + "\n\n"
        )


        # Column headings
        self.result_text.insert(
            tk.END,
            f"{'':>8}"
        )


        for vertex in vertices:

            self.result_text.insert(
                tk.END,
                f"{vertex:>10}"
            )


        self.result_text.insert(
            tk.END,
            "\n"
        )


        self.result_text.insert(
            tk.END,
            "-" * (
                8 + 10 * len(vertices)
            )
            + "\n"
        )


        # Matrix rows
        for i in range(
            len(vertices)
        ):

            self.result_text.insert(
                tk.END,
                f"{vertices[i]:>8}"
            )


            for j in range(
                len(vertices)
            ):

                value = matrix[i][j]


                if value == INF:

                    text = "∞"

                else:

                    text = self.format_number(
                        value
                    )


                self.result_text.insert(
                    tk.END,
                    f"{text:>10}"
                )


            self.result_text.insert(
                tk.END,
                "\n"
            )


    # ========================================================
    # FORMAT NUMBER
    # ========================================================

    def format_number(
        self,
        value
    ):

        if float(value).is_integer():

            return str(
                int(value)
            )


        return str(
            round(
                value,
                2
            )
        )


    # ========================================================
    # SHOW GRAPH
    # ========================================================

    def show_graph(self):

        self.canvas.delete(
            "all"
        )


        vertices = list(
            self.graph.keys()
        )


        # No vertices
        if len(vertices) == 0:

            self.canvas.create_text(
                350,
                170,
                text="Add vertices to display graph",
                font=("Segoe UI", 14),
                fill="#94A3B8"
            )

            return


        self.canvas.update()


        width = self.canvas.winfo_width()

        height = self.canvas.winfo_height()


        if width < 400:

            width = 600


        if height < 250:

            height = 350


        center_x = width / 2

        center_y = height / 2


        radius = min(
            width,
            height
        ) * 0.35


        positions = {}


        # ====================================================
        # POSITION VERTICES IN CIRCLE
        # ====================================================

        for i, vertex in enumerate(
            vertices
        ):

            angle = (
                2
                * math.pi
                * i
                / len(vertices)
            )


            x = (
                center_x
                + radius
                * math.cos(angle)
            )


            y = (
                center_y
                + radius
                * math.sin(angle)
            )


            positions[vertex] = (
                x,
                y
            )


        
        # DRAW UNDIRECTED EDGES ONCE
        drawn_edges = set()

        for source in self.graph:
            for destination, weight in self.graph[source].items():
                edge_key = frozenset((source, destination))
                if edge_key in drawn_edges:
                    continue
                drawn_edges.add(edge_key)

                x1, y1 = positions[source]
                x2, y2 = positions[destination]

                self.canvas.create_line(
                    x1, y1, x2, y2,
                    fill="#8A9AAA",
                    width=2
                )

                mx = (x1 + x2) / 2
                my = (y1 + y2) / 2

                self.canvas.create_oval(
                    mx - 18, my - 11, mx + 18, my + 11,
                    fill=WHITE,
                    outline="#CBD5E1"
                )

                self.canvas.create_text(
                    mx, my,
                    text=self.format_number(weight),
                    font=("Segoe UI", 9, "bold"),
                    fill=TEXT_COLOR
                )


        # ====================================================
        # DRAW VERTICES
        # ====================================================

        for vertex in vertices:

            x, y = positions[
                vertex
            ]


            self.canvas.create_oval(
                x - 28,
                y - 28,
                x + 28,
                y + 28,
                fill=PRIMARY_COLOR,
                outline=WHITE,
                width=3
            )


            self.canvas.create_text(
                x,
                y,
                text=vertex,
                font=(
                    "Segoe UI",
                    12,
                    "bold"
                ),
                fill=WHITE
            )



    # CLEAR INPUT
   

    def clear_input(self):

        # Vertex input
        self.vertex_entry.delete(
            0,
            tk.END
        )


        # Edge inputs
        self.edge_source.delete(
            0,
            tk.END
        )

        self.edge_destination.delete(
            0,
            tk.END
        )

        self.edge_weight.delete(
            0,
            tk.END
        )


        # Algorithm inputs
        self.source_entry.delete(
            0,
            tk.END
        )

        self.destination_entry.delete(
            0,
            tk.END
        )


  
    # CLEAR ALL
 

    def clear_all(self):

        answer = messagebox.askyesno(
            "Clear All",
            "Are you sure you want to delete the complete graph?"
        )


        if answer:

            self.graph.clear()


            self.vertex_listbox.delete(
                0,
                tk.END
            )


            self.canvas.delete(
                "all"
            )


            self.result_text.delete(
                "1.0",
                tk.END
            )


            self.clear_input()


            self.show_graph()



# START PROGRAM


if __name__ == "__main__":

    root = tk.Tk()

    app = ShortestPathApp(
        root
    )

    root.mainloop()
```
